```python
import pandas as pd

X = pd.read_csv("/Users/pareesoni/Downloads/archive/X_train_2025.csv")
y = pd.read_csv("/Users/pareesoni/Downloads/archive/y_train_2025.csv")

print(X.shape)
print(y.shape)
```

```python
print(X.columns)
print(y.head())
print(X.head())
```

```python
print(y['In-hospital_death'].value_counts())
```

```python
X.isnull().sum().sort_values(ascending=False).head(10)
```

```python
# Drop columns with too many missing values (>70%)

threshold = 0.7 * len(X)

X = X.dropna(axis=1, thresh=threshold)

print(X.shape)
```

```python
# Fill missing values with median

X = X.fillna(X.median())

# Check if any missing values remain
print(X.isnull().sum().sum())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y['In-hospital_death'], test_size=0.2, random_state=42
)

print(X_train.shape)
print(X_test.shape)
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(max_iter=2000, solver='liblinear')
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)

print(y_pred[:10])
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred))
```

```python
model = LogisticRegression(max_iter=2000, solver='liblinear', class_weight='balanced')
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred))
```

```python
y_prob = model.predict_proba(X_test)[:, 1]

print(y_prob[:10])
```

```python
import numpy as np

# Set threshold (you can tune this)
threshold = 0.5

y_pred_custom = (y_prob > threshold).astype(int)

from sklearn.metrics import classification_report
print(classification_report(y_test, y_pred_custom))
```

```python
threshold = 0.3

y_pred_custom = (y_prob > threshold).astype(int)

from sklearn.metrics import classification_report
print(classification_report(y_test, y_pred_custom))
```

```python
from sklearn.metrics import accuracy_score

print(accuracy_score(y_test, y_pred_custom))
```

```python
from xgboost import XGBClassifier

xgb_model = XGBClassifier(eval_metric='logloss')
xgb_model.fit(X_train, y_train)
```

```python
y_pred_xgb = xgb_model.predict(X_test)
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred_xgb))
```

```python
xgb_model = XGBClassifier(
    eval_metric='logloss',
    scale_pos_weight=6  # important
)

xgb_model.fit(X_train, y_train)
```

```python
y_pred_xgb = xgb_model.predict(X_test)

from sklearn.metrics import classification_report
print(classification_report(y_test, y_pred_xgb))
```

```python
import numpy as np

importance = model.coef_[0]

feature_names = X.columns

# Get top 10 important features
indices = np.argsort(np.abs(importance))[-10:]

for i in indices:
    print(feature_names[i], importance[i])
```

```python
import pandas as pd
from sklearn.linear_model import LogisticRegression
import joblib

# Fake sample dataset (for demo)
data = pd.DataFrame({
    "age": [25, 70, 60, 45, 80],
    "heart_rate": [70, 110, 95, 85, 120],
    "bp": [120, 150, 140, 130, 160],
    "target": [0, 1, 1, 0, 1]
})

X = data[["age", "heart_rate", "bp"]]
y = data["target"]

model = LogisticRegression()
model.fit(X, y)

# Save model
joblib.dump(model, "model.pkl")

print("Model saved ✅")
```
